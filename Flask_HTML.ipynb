# Introduction to Flask
Flask is a lightweight web framework written in Python. It's designed to make getting started with web development easy, while also being flexible enough to serve as the foundation for large applications.

## Key features:

Minimalistic and easy to get started
Extensibility to add more features as your application grows
Built-in development server 
RESTful request dispatching
Uses Jinja2 templating

# Jinja2 
### is a powerful templating engine for Python, widely used in web frameworks like Flask and Django. It allows you to embed dynamic content and logic directly into HTML files using special syntax.

## Key features of Jinja2 include:

### Variable Interpolation: You can insert Python variables into your HTML using double curly braces, like {{ result }}. This replaces the placeholder with the actual value when the page is rendered.

### Control Structures: Jinja2 supports logic such as conditionals ({% if ... %}), loops ({% for ... %}), and more. This lets you show or hide parts of your page, repeat elements, or handle complex rendering scenarios.

### Filters and Expressions: You can modify variables with filters, such as {{ name|upper }} to convert a string to uppercase.
